```python
def calculator():
    print("=== 간단한 파이썬 계산기 ===")
    try:
        num1 = float(input("첫 번째 숫자를 입력하세요: "))
        operator = input("연산자를 입력하세요 (+, -, *, /): ")
        num2 = float(input("두 번째 숫자를 입력하세요: "))

        if operator == '+':
            result = num1 + num2
        elif operator == '-':
            result = num1 - num2
        elif operator == '*':
            result = num1 * num2
        elif operator == '/':
            if num2 == 0:
                return "오류: 0으로 나눌 수 없습니다."
            result = num1 / num2
        else:
            return "오류: 잘못된 연산자입니다."

        return f"결과: {num1} {operator} {num2} = {result}"
    except ValueError:
        return "오류: 올바른 숫자를 입력해주세요."

# 계산기 실행
print(calculator())
```

```python
# 1. 정수형 (int)
a = 123
print(type(a))  # <class 'int'> 출력 (소수점이 없는 정수)

# 2. 실수형 (float)
a = 123.456
print(type(a))  # <class 'float'> 출력 (소수점이 포함된 실수)

# 3. 문자열 (str)
a = "Hello World!"
print(type(a))  # <class 'str'> 출력 (큰따옴표/작은따옴표로 감싼 문자열)

# 4. 불리언/논리형 (bool)
a = True
print(type(a))  # <class 'bool'> 출력 (참/거짓을 나타내는 논리값)

# 5. 리스트 (list)
a = [1, 2, 3]
print(type(a))  # <class 'list'> 출력 (대괄호로 묶인 순서가 있는 데이터 집합)
```

```python
# 1. True(1) + False(0) 연산
a = True; b = False
print(a+b)  # 1 + 0 = 1 출력

# 2. 덧셈 연산
a = 3; b = 4
c = a + b
print(c)  # 7 출력

# 3. 크기 비교 (3 < 4)
d = a < b
print(d)  # 참이므로 True 출력

# 4. 양쪽 다 참인지 확인 (AND 연산)
e = 3 < 4 and 4 < 5
print(e)  # 둘 다 참이므로 True 출력
```
